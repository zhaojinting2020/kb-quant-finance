本节主要介绍如何使用 Python 读取金融数据：

*原生Python 

*csv.reader

*pandas

pandas 在数据读取、处理、存储方面更高效。

```python
# 如果下载失败，可以打开代理
# import os
# proxy = 'http://127.0.0.1:7897'
# os.environ['HTTP_PROXY'] = proxy 
# os.environ['HTTPS_PROXY'] = proxy 

import yfinance as yf

symbol = 'AAPL'
fn = 'AAPL.csv'

start_date = '2026-02-01'
end_date = '2026-03-01'

df = yf.download(symbol, start=start_date, end=end_date)

if not df.empty:
    df.to_csv(fn)
    print(f"数据已保存: {fn}")
else:
    print("数据获取失败")
```

```python
# 用原生 Python 

with open(fn, 'r') as f:
    for _ in range(6):
        print(f.readline(), end='')
```

```python
# 使用 csv.reader

import csv
csv_reader = csv.reader(open(fn, 'r'))
data = list(csv_reader)
data[:6]
```

```python
# 使用 csv.DictReader

csv_reader = csv.DictReader(open(fn, 'r'))
data = list(csv_reader)
data[:3]
```

```python
data_clean = data[2:]  # 跳过前两行
sum([float(l['Close']) for l in data_clean]) / len(data_clean)
```

```python
# 使用 pandas

import pandas as pd
data = pd.read_csv(
    'AAPL.csv',
    skiprows=[1, 2],   # 跳过脏行
    index_col=0
)

# 突出金融数据的时间序列属性，体现pandas的优势
data.index = pd.to_datetime(data.index)
data.sort_index(inplace=True)

data.head()
```

本节对比了三种数据读取方式：

*原生 Python：只能读，不能分析

*csv 模块：结构稍好，但操作繁琐

*pandas：结构清晰，计算高效

实际金融数据往往包含脏数据（如多余表头），需要清洗后才能使用。pandas 通过 DataFrame 提供统一结构，并天然支持时间序列处理（索引、排序、滚动计算等）。
